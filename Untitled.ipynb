```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Define the data
data = {
    'Dimension': ['Validity', 'Completeness', 'Integrity', 'Uniqueness', 'Timeliness', 'Reasonableness'],
    'DR1': [92.0, 88.5, 89.0, 86.0, 83.0, 84.0],
    'DR2': [97.5, 96.0, 98.0, 94.0, 91.0, 93.0],
}

overall_scores = {
    'DR1': 96.2,
    'DR2': 99.0
}

# Step 2: Convert to DataFrame
df = pd.DataFrame(data)

# Step 3: Setup the figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Step 4: Bar chart settings
bar_width = 0.4
index = np.arange(len(df['Dimension']))

# Bars for DR1 and DR2
bars_dr1 = ax.bar(index - bar_width / 2, df['DR1'], bar_width, label='DR1', color='lightblue')
bars_dr2 = ax.bar(index + bar_width / 2, df['DR2'], bar_width, label='DR2', color='royalblue')

# Step 5: Add labels and title
ax.set_title(f"Data Quality (DQ) Scorecard: DR1 vs DR2\nOverall DQ Pass Rates: DR1 = {overall_scores['DR1']}%, DR2 = {overall_scores['DR2']}%", fontsize=14, pad=20)
ax.set_xlabel("DQ Dimensions", fontsize=12)
ax.set_ylabel("DQ Score (%)", fontsize=12)
ax.set_xticks(index)
ax.set_xticklabels(df['Dimension'], fontsize=11)
ax.set_ylim(0, 105)  # Extend Y-axis slightly above 100% to prevent label overlap



# Step 7: Annotate scores on the bars (adjust placement)
for bars, scores in zip([bars_dr1, bars_dr2], [df['DR1'], df['DR2']]):
    for bar, score in zip(bars, scores):
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 2, f"{score:.1f}%", 
                ha='center', va='bottom', fontsize=10, color='black')  # Place numbers above the bars

# Step 8: Add a subtitle for overall pass values
fig.suptitle(
    f"Overall DQ Pass Rates\nDR1: {overall_scores['DR1']}% | DR2: {overall_scores['DR2']}%", 
    x=0.5, y=0.02, fontsize=12, ha="center", fontweight="bold"
)

# Step 9: Save and show the figure
plt.tight_layout(rect=[0, 0.05, 1, 0.95])  # Add space for the subtitle
plt.savefig("figure_dq1_quality_scorecard_legend_fix.png", dpi=300)
plt.show()
```
