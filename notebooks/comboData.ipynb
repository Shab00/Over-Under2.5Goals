```python
folder_path = '../data/raw/csvFiles'
combined_csv_path = 'combined.csv'
csv_list = []

folder_path = '../data/raw/csvFiles'
combined_csv_path = 'combined.csv'
csv_list = []

columns_to_keep = [
    'Div', 'Date', 'Time', 'HomeTeam', 'AwayTeam', 'FTHG', 'FTAG', 'FTR',
    'HTHG', 'HTAG', 'HTR', 'Attendance', 'Referee', 'HS', 'AS', 'HST', 'AST',
    'HHW', 'AHW', 'HC', 'AC', 'HF', 'AF', 'HFKC', 'AFKC', 'HO', 'AO', 'HY', 'AY',
    'HR', 'AR', 'IWH', 'IWD', 'IWA', 'WHH', 'WHD', 'WHA'
]

for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)
        try:
            df = pd.read_csv(file_path, encoding='latin1', on_bad_lines='skip')
            df_filtered = df[[col for col in columns_to_keep if col in df.columns]]
            csv_list.append(df_filtered)
        except Exception as e:
            print(f"Error reading {file_path}: {e}")

if csv_list:
    combined_df = pd.concat(csv_list, ignore_index=True)
    combined_df.to_csv(combined_csv_path, index=False)
    print(f"Combined CSV saved to {combined_csv_path}")
else:
    print("No CSV files were combined.")
```
