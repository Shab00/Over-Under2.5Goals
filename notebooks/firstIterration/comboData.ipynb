# Football Match Data ML Project – Iteration 1

## 1. Introduction

This notebook documents the first iteration of my machine learning project using football match data.  
The primary goal in this stage is to aggregate historical match data from multiple CSV files into a single, clean dataset for future analysis and modeling.

---

## 2. Data Collection and Combination

- All raw CSV data files are stored in the directory: `../data/raw/csvFiles`
- Each file contains data for one or more football seasons or competitions.
- Only a specific set of columns relevant for match outcomes and betting odds are retained in the combined dataset.

**Steps:**
1. Iterate through all `.csv` files in the raw data folder.
2. For each file:
    - Read the CSV using pandas, handling encoding and skipping bad lines.
    - Keep only columns in the predefined `columns_to_keep` list.
    - Append the filtered DataFrame to a list.
3. Concatenate all DataFrames from the list into a single DataFrame.
4. Save the combined data to `combinedWithOdds.csv` for use in later steps.

---

## 3. Code Summary

```python
import pandas as pd
import os

folder_path = '../data/raw/csvFiles'
combined_csv_path = 'combinedWithOdds.csv'
csv_list = []

columns_to_keep = [
    # [list of all columns to keep, as in your code above]
]

for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)
        try:
            df = pd.read_csv(file_path, encoding='latin1', on_bad_lines='skip')
            df_filtered = df[[col for col in columns_to_keep if col in df.columns]]
            csv_list.append(df_filtered)
        except Exception as e:
            print(f"Error reading {file_path}: {e}")

if csv_list:
    combined_df = pd.concat(csv_list, ignore_index=True)
    combined_df.to_csv(combined_csv_path, index=False)
    print(f"Combined CSV saved to {combined_csv_path}")
else:
    print("No CSV files were combined.")
```
